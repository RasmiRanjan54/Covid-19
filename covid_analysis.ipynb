```python
df = df.drop_duplicates()
```

```python
df['DATE_DIED'] = df['DATE_DIED'].replace('9999-99-99', pd.NaT)

```

```python
df['DATE_DIED'] = pd.to_datetime(df['DATE_DIED'], errors='coerce')
df['DEATH'] = df['DATE_DIED'].notna().astype(int)

```

```python
df = df[df['SEX'].isin([1, 2])]
```

```python
df['SEX'] = df['SEX'].replace({1: 'Female', 2: 'Male'})
```

```python
sns.displot(df['AGE'])
```

```python
import matplotlib.pyplot as plt

death_by_gender = df[df['DEATH'] == 1]['SEX'].value_counts()

death_by_gender.plot(kind='bar')

plt.title('COVID Death Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Deaths')

plt.show()
```

```python
categ_cols = [
    'USMER', 'MEDICAL_UNIT', 'SEX', 'PATIENT_TYPE', 'PNEUMONIA',
    'PREGNANT', 'DIABETES', 'COPD', 'ASTHMA', 'INMSUPR',
    'HIPERTENSION', 'OTHER_DISEASE', 'CARDIOVASCULAR', 'OBESITY',
    'RENAL_CHRONIC', 'TOBACCO', 'CLASIFFICATION_FINAL'
]
```

```python
# Create DIED column correctly
df['DIED'] = df['DATE_DIED'].notna().astype(int)

# Define color palette
palette = {0: 'green', 1: 'red'}

# Plot categorical columns
for col in categ_cols:
    plt.figure(figsize=(10,6))
    sns.countplot(data=df, x=col, hue="DIED", palette=palette)

    plt.title(f'{col} Distribution by Mortality (DIED)', fontsize=16, weight='bold')
    plt.xlabel(col, fontsize=13)
    plt.ylabel('Count', fontsize=13)

    plt.xticks(rotation=45)
    plt.legend(title="DIED (0 = Alive, 1 = Died)")
    plt.tight_layout()
    plt.show()
```

```python
plt.figure(figsize=(10,7))
corr = df[['AGE','PNEUMONIA','DIABETES','HIPERTENSION','OBESITY','DIED']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=.5, fmt='.2f')
plt.title('Correlation Between Key COVID-19 Variables', fontsize=16, weight='bold')
plt.tight_layout()
plt.show()
```
